# Running Llama on Astraflow

[Astraflow](https://astraflow.ucloud-global.com) by UCloud is an OpenAI-compatible AI model aggregation platform supporting 200+ models.

Because Astraflow exposes an OpenAI-compatible API, the only change needed compared to using the OpenAI SDK directly is setting `base_url` and `api_key`.

**Endpoints**
| Region | Base URL | Env var |
|--------|----------|---------|
| Global | `https://api-us-ca.umodelverse.ai/v1` | `ASTRAFLOW_API_KEY` |
| China  | `https://api.modelverse.cn/v1`        | `ASTRAFLOW_CN_API_KEY` |

Sign up for a free API key at https://astraflow.ucloud-global.com (global) or https://astraflow.ucloud.cn (China).

## 2. Configure the client

```python
import os
from openai import OpenAI

# For the China endpoint, use:
#   api_key=os.environ['ASTRAFLOW_CN_API_KEY']
#   base_url='https://api.modelverse.cn/v1'

client = OpenAI(
    api_key=os.environ['ASTRAFLOW_API_KEY'],
    base_url='https://api-us-ca.umodelverse.ai/v1',
)
```

## 3. Basic chat completion

```python
response = client.chat.completions.create(
    model='meta-llama/Llama-3.3-70B-Instruct',
    messages=[
        {'role': 'system', 'content': 'You are a helpful assistant.'},
        {'role': 'user', 'content': 'What are the key features of Meta Llama 3.3?'},
    ],
    max_tokens=512,
)
print(response.choices[0].message.content)
```

## 4. Streaming

```python
stream = client.chat.completions.create(
    model='meta-llama/Llama-3.3-70B-Instruct',
    messages=[
        {'role': 'system', 'content': 'You are a helpful assistant.'},
        {'role': 'user', 'content': 'Write a short poem about open-source AI.'},
    ],
    max_tokens=256,
    stream=True,
)
for chunk in stream:
    delta = chunk.choices[0].delta
    if delta.content:
        print(delta.content, end='', flush=True)
```

## 5. Trying other Llama models

Astraflow supports 200+ models. You can swap the `model` parameter to use different Llama variants:

```python
# Example model IDs available on Astraflow (check the platform for the full list):
LLAMA_MODELS = [
    'meta-llama/Llama-3.3-70B-Instruct',
    'meta-llama/Llama-3.1-8B-Instruct',
    'meta-llama/Llama-3.1-70B-Instruct',
]

question = 'In one sentence, explain what a large language model is.'

for model_id in LLAMA_MODELS:
    resp = client.chat.completions.create(
        model=model_id,
        messages=[{'role': 'user', 'content': question}],
        max_tokens=128,
    )
    print(f'{model_id}:\n  {resp.choices[0].message.content}\n')
```
